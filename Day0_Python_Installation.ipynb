| Escape Sequence | Description                               | Example Output                   |
| --------------- | ----------------------------------------- | -------------------------------- |
| `\\`            | Backslash                                 | `\\` → `\`                       |
| `\'`            | Single quote                              | `\'` → `'`                       |
| `\"`            | Double quote                              | `\"` → `"`                       |
| `\n`            | Newline (line break)                      | `Hello\nWorld` → on 2 lines      |
| `\t`            | Tab (horizontal space)                    | Adds spacing like pressing Tab   |
| `\r`            | Carriage return (return to line start)    | `ABC\r123` → `123`               |
| `\f`            | Form feed (new page in printing)          | Rare, depends on system          |
| `\v`            | Vertical tab                              | Rare, shows vertical spacing     |
| `\a`            | Bell/Alert (sound)                        | May make beep sound in terminal  |
| `\b`            | Backspace (moves one char back)           | `abc\b` → `ab`                   |

### Rules for Defining Variables:
1. **Must start with a letter or an underscore (_)** (e.g., `name`, `_value`).
2. **Cannot start with a number** (e.g., `2value` is not allowed).
3. **Can contain letters, numbers, and underscores** (e.g., `value2`, `total_sum`).
4. **Case-sensitive** (e.g., `Age` and `age` are different variables).
5. **Cannot use Python reserved keywords** (e.g., `for`, `if`, `while`).

**String Formating in Python** - To print the statement along with variable values inside print(f" name is {}")
- 1. Using f-strings (Python 3.6+)
- 2. Using the str.format() Method
- 3. Using % Formatting

```python
# 1. Using f-strings
V= 10
# print(V)
# print("The value of V is",V)
print(f"{V} is the value of V")
```

```python
name = "SaiGanesh"
print(f"Name is {name}, welcome to the Algonex community!")
```

```python
#1. Using f-strings  - To print the statement along with variable values inside print(f" name is {}")
name = "Karlapoodi"
age = 30
print(f"Name is {name}, Age is {age}")
print(age,name)
print("My Age and Name is",age,name)
```

```python
#2. Using the str.format() Method - first {} will be replaced by first variable and second {} by second variable
name = "Algonex"
age = 5
print("Name: {}, Age: {}".format(name, age))

# print("Name: {}, Age: {}".format(age, name))
```

```python
#3. Using % Formatting - %s is for string and %d is for integer
name = "Carol"
age = 22
print("Name: %s, Age: %d" % (name, age))
```
