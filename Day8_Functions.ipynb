```python
# Function to apply discount
def apply_discount(total, discount_percent):
    return total - (total * discount_percent / 100)

# Function to take an order and apply discount if needed
def take_order(name, drink, price_per_drink, quantity):
    total = price_per_drink * quantity
    if quantity >= 3:
        total = apply_discount(total, 10)  # 10% discount
        print(f"Congrats {name}, you got a 10% discount!")
    else:
        print(f"Thank you {name},OOPS!! this time no discount applied.")
    print(f"{name}, your total for {quantity} {drink}(s) is: ${total:.3f}")

# Test with different customers
take_order("Bellamkonda", "Latte", 4.5, 2)   # No discount
take_order("Cheepineti", "Cappuccino", 5.0, 3) # Discount applied
take_order("Santenna", "Espresso", 3.0, 5) # Discount applied
```

```python
def apply_discount(total, discount_percent):
    """
    Returns the total after applying a discount percentage.
    """
    discount = total * (discount_percent / 100)
    return total - discount

def take_order(customer, drink, quantity, price_per_drink):
    """
    Calculates the total bill for an order. 
    Applies a 10% discount if quantity is 3 or more.
    """
    total = quantity * price_per_drink
    discount_applied = False
    if quantity >= 3:
        total = apply_discount(total, 10)
        discount_applied = True
    return total, discount_applied

# Prices of drinks
prices = {
    'Latte': 5.0,
    'Espresso': 4.5,
    'Cappuccino': 5.0,
    'Mocha': 5.0
}

# Test cases
orders = [
    ("Alice", "Latte", 2),
    ("Bob", "Espresso", 3),
    ("Charlie", "Cappuccino", 5),
    ("Donna", "Mocha", 1)
]

# Process orders
for customer, drink, quantity in orders:
    price = prices[drink]
    total, discount_applied = take_order(customer, drink, quantity, price)
    print(f"{customer} ordered {quantity} {drink}(s). "
          f"{'Discount applied. ' if discount_applied else ''}"
          f"Total amount: ${total:.2f}")
```

## Mini Project: Student Marks Management System (Using Only Functions)
#### Objective

Build a simple system that can:

- Add student details

- Calculate total & average marks

- Display results

- Work fully using only functions

**Concepts Covered**

- Functions with parameters
- Return values
- Default arguments
- Loops inside functions
- Menu-driven program
- Real-life application logic

**Features**

- Add student details
- Calculate total marks
- Calculate average
- Display result (Pass/Fail)
- Menu-driven interaction
- Fully function-based design

```python
# Mini Project: Student Marks Management System - Using ONLY Functions

def add_student():
    name = input("Enter Student Name: ")
    m1 = int(input("Enter Marks 1: "))
    m2 = int(input("Enter Mks 2: "))
    m3 = int(input("Enter Maks 3: "))
    return name, m1, m2, m3

def calculate_total(m1, m2, m3):
    return m1 + m2 + m3

def calculate_average(total):
    return total / 3

def calculate_result(m1, m2, m3):
    if m1 >= 35 and m2 >= 35 and m3 >= 35:
        return "PASS"
    else:
        return "FAIL"

def display_report(name, m1, m2, m3, total, avg, result):
    print("\n------ STUDENT REPORT ------")
    print("Name       :", name)
    print("Marks      :", m1, m2, m3)
    print("Total      :", total)
    print("Average    :", avg)
    print("Result     :", result)
    print("----------------------------")

def main():
    while True:
        print("\n--- STUDENT MARKS SYSTEM ---")
        print("1. Add Student & Generate Report")
        print("2. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            name, m1, m2, m3 = add_student()
            total = calculate_total(m1, m2, m3)
            avg = calculate_average(total)
            result = calculate_result(m1, m2, m3)
            display_report(name, m1, m2, m3, total, avg, result)

        elif choice == "2":
            print("Thank you for using the system!")
            break

        else:
            print("Invalid choice! Try again.")

# Program starts here
main()
```
