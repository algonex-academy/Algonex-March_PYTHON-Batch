# 2. while Loop

Used when you don't know how many times the loop will run, but it should continue until a condition becomes false.

```python
while condition:
    # statements
```

- As long as the condition is TRUE, loop keeps running.

- Stop only when condition becomes FALSE.

```python
# Example 1: Print numbers 1–5
i = 1
while i <= 5:
    print(i)
    i += 1
```

```python
# step-up while condition values (1->5)

num = 10
while num <= 20:
          print(num)
          num+=1 
          # num+= 2   2-step up values
          
```

```python
# step-down while condition values ( 5-> 1)
n = 5
while n>0:
          print(n)
          n= n-1   # n-=1
          # n-=2 for two step-down values
```

```python
# Example 2: ATM PIN attempt (industry: banking security)
pin = "1234"
user_pin = input("Enter PIN: ")

while user_pin != pin:
    print("Wrong PIN! Try again.")
    user_pin = input("Enter PIN: ")
   

print("Access Granted!")
```

```python
# Example 3: Retry payment until success (industry: e-commerce)
success = False

while not success:
    status = input("Payment successful? (yes/no): ")

    if status == "yes":
        success = True

print("Order Confirmed!")
```

### 3. **break Statement**

Used to stop the loop immediately, even if items are left.

```python
# Example (Stop when error occurs)
statuses = ["ok", "ok", "error", "ok"]

for s in statuses:
    if s == "error":
        print("Error detected! Stopping process.")
        break
    print("Processing:", s)
```

**Industry use case:**

Stopping a file upload, stopping a payment, stopping data processing


### 4. continue Statement

Skips the current iteration and moves to the next one.

```python
# Example (Skip invalid data)
values = [10, -3, 20, -8, 30]

for v in values:
    if v < 0:
        continue
    print("Accepted value:", v)
```

**Industry Use:**

Skipping blank entries, skipping negative numbers, skipping invalid records.

| Loop Type      | When to Use                                | Example Industry Use               |
| -------------- | ------------------------------------------ | ---------------------------------- |
| **for loop**   | When you know number of iterations         | Process list of customers/products |
| **while loop** | When you don’t know how long it should run | Login attempts, retry payments     |
| **break**      | To stop loop early                         | Stop when error found              |
| **continue**   | Skip current iteration                     | Skip invalid data                  |

Build a Python program where:

Project Description

Build a Python program where:

1. The system asks for the number of students

2. Repeatedly collects attendance (“P” for Present, “A” for Absent)

Calculates:

- Total Present

- Total Absent
- Attendance Percentage
- Displays results as a simple report

```python
print("===== Student Attendance Tracker =====")

num_students = int(input("Enter total number of students: "))

present_count = 0
absent_count = 0

for i in range(1, num_students + 1):
    status = input(f"Enter attendance for student {i} (P/A): ").upper()
    
    while status not in ["P", "A"]:
        print("Invalid input! Please enter P or A only.")
        status = input(f"Enter attendance for student {i} (P/A): ").upper()
    
    if status == "P":
        present_count += 1
    else:
        absent_count += 1

attendance_percentage = (present_count / num_students) * 100

print("\n--- Attendance Report ---")
print("Total Students:", num_students)
print("Present:", present_count)
print("Absent:", absent_count)
print("Attendance %:", round(attendance_percentage, 2))
print("--------------------------")
```

### Mini Real-Time Project

Student Marks Processing System (using loops)
Objective:

- Collect marks from students
- Skip invalid entries
- Calculate total, average, highest, lowest

```python
print("=== Student Marks Processing System ===")

num = int(input("Enter number of students: "))
marks = []

for i in range(num):
    m = int(input(f"Enter marks of student {i+1}: "))

    if m < 0 or m > 100:
        print("Invalid marks! Skipping entry...")
        continue

    marks.append(m)

print("\n--- Report ---")
print("Marks:", marks)
print("Total:", sum(marks))
print("Average:", sum(marks) / len(marks))
print("Highest:", max(marks))
print("Lowest:", min(marks))
```
