

### **1. Variables by Scope (Be focus)**
The scope of a variable determines where it can be accessed, Where it cannot be accessed within a program. 

*Scope decides the “visibility” of a variable in a program*

Why Variable Scope is Important in Industry

In real projects:
- It avoids data conflicts
- It improves security
- It reduces bugs
- It makes code clean and maintainable
- It controls memory usage

Python has four main variable scopes:

#### **a. Local Variables**
- **Definition**: Variables defined inside a function or block, accessible only within that function or block.

*Meaning is Temporary data used only for one specific task and Exists only during the function’s execution*
- **Example**:
  ```python
  def my_function():
      local_var = 10  # Local variable
      print(local_var)

  my_function()  # Output: 10
  # print(local_var)  # Error: NameError, local_var is not defined outside the function


  ### Another example

  def calculate_salary():
    bonus = 5000   # Local variable
    print(bonus) # 5000
  calculate_salary()
  print(bonus)  -- Error (outside function)


def generate_payslip():
    tax = 2000   # Only used inside payslip
    print("Tax:", tax)

generate_payslip()

Used in:

- Payroll systems
- Billing systems
- Login validation
- Temporary calculations

### **b. Global Variables**
 Variables defined outside any function or class, accessible throughout the module

*A global variable is declared outside all functions and can be used anywhere in the program* and Exists for the entire duration of the program (unless explicitly deleted)

Example:
global_var = 20  # Global variable

def my_function():
    print(global_var) #20

my_function()  # Output: 20
print(global_var)  # Output: 20
Modifying Global Variables: Use the global keyword inside a function to modify a global variable

global_var = 20

def modify_global():
    global global_var
    global_var = 30

modify_global()
print(global_var)  # Output: 30

company_name = "Infosys"   # Global variable

def display_company():
    print(company_name)

display_company()

bank_name = "SBI"

def show_bank():
    print("Welcome to", bank_name)

show_bank()

Used in:

- App name
- Company name
- Base URL
- Database connection string

### Changing Global Variable Inside a Function

By default, Python does NOT allow modifying a global variable inside a function unless you use the global keyword

  

Industry Use:
- Update total users

- Update counter values



c. Enclosing (Nonlocal) Variables
**Variables defined in an enclosing function (outer function in nested functions), accessible to inner functions and Exists while the enclosing function is active**

Used when you have:

- A function inside another function
- And you want to modify the outer function’s variable

Example:
def outer_function():
    enclosing_var = "I am enclosing"

    def inner_function():
        print(enclosing_var)  # Accessing enclosing variable
    inner_function()

outer_function()  # Output: I am enclosing

#### Modifying Enclosing Variables: Use the nonlocal keyword to modify a variable in the enclosing scope

def outer_function():
    enclosing_var = "Original"

    def inner_function():
        nonlocal enclosing_var
        enclosing_var = "Modified"
        print(enclosing_var)
    inner_function()
    print(enclosing_var)

outer_function()  # Modified

#### Example -2
def outer():
    count = 0

    def inner():
        nonlocal count
        count = count+1
        print(count) # 1

    inner() 
    print(count) # 1

d. Built-in Variables
**Variables provided by Python’s built-in namespace, such as True, False, None, or built-in module names and Available everywhere unless shadowed by a variable with the same name**

These are default Python functions available everywhere:
- print()
- len()
- input()
- type()
- sum()

#### These work in all programs without defining them


Example:
print(True)  # Built-in variable
# Shadowing a built-in
Tru = "Not a boolean anymore"  # Avoid this in practice!
print(Tru)  # Output: Not a boolean anymore 



## LEGB Rule
Python resolves variable names using the LEGB rule (order of lookup):

Local: Inside the current function
Enclosing: In outer functions (for nested functions)
Global: In the module’s global scope
Built-in: In Python’s built-in namespace


### **2. Variables by Data Type**
Python variables can hold different data types, which are determined by the value assigned to them. Here are the primary data types associated with variables:

#### **a. Numeric Types**
- **Integer (`int`)**: Whole numbers, e.g., `42`, `-10`.
- **Float (`float`)**: Decimal numbers, e.g., `3.14`, `-0.001`.
- **Complex (`complex`)**: Numbers with real and imaginary parts, e.g., `3 + 4j`.

- **Example**:
  ```python
  int_var = 42
  float_var = 3.14
  complex_var = 3 + 4j
  print(type(int_var), type(float_var), type(complex_var))
  # Output: <class 'int'> <class 'float'> <class 'complex'>
  ```

#### **b. Sequence Types**
- **String (`str`)**: Sequence of characters, e.g., `"hello"`, `'Python'`.
- **List (`list`)**: Ordered, mutable collection, e.g., `[1, 2, 3]`.
- **Tuple (`tuple`)**: Ordered, immutable collection, e.g., `(1, 2, 3)`.

- **Example**:
  ```python
  str_var = "Hello, Python"
  list_var = [1, "apple", 3.14]
  tuple_var = (10, 20, 30)
  print(str_var, list_var, tuple_var)
  # Output: Hello, Python [1, 'apple', 3.14] (10, 20, 30)
  ```

#### **c. Mapping Type**
- **Dictionary (`dict`)**: Key-value pairs, e.g., `{"name": "Alice", "age": 25}`.
- **Example**:
  ```python
  dict_var = {"name": "Alice", "age": 25}
  print(dict_var["name"])  # Output: Alice
  ```

#### **d. Set Types**
- **Set (`set`)**: Unordered collection of unique elements, e.g., `{1, 2, 3}`.
- **Frozenset (`frozenset`)**: Immutable version of a set, e.g., `frozenset([1, 2, 3])`.

- **Example**:
  ```python
  set_var = {1, 2, 3, 3}  # Duplicates are removed
  frozenset_var = frozenset([1, 2, 3])
  print(set_var, frozenset_var)
  # Output: {1, 2, 3} frozenset({1, 2, 3})
  ```

#### **e. Boolean Type**
- **Boolean (`bool`)**: Represents `True` or `False`.
- **Example**:
  ```python
  bool_var = True
  print(bool_var, type(bool_var))
  # Output: True <class 'bool'>
  ```

#### **f. None Type**
- **NoneType (`None`)**: Represents the absence of a value, similar to `null` in other languages.
- **Example**:
  ```python
  none_var = None
  print(none_var, type(none_var))
  # Output: None <class 'NoneType'>
  ```

#### **g. Custom Objects**
- Variables can hold instances of user-defined classes.
- **Example**:
  ```python
  class Person:
      def __init__(self, name):
          self.name = name

  person_var = Person("Alice")
  print(person_var.name)  # Output: Alice
  ```

### **3. Special Variable Types**
These are variables with specific behaviors or conventions in Python:

#### **a. Instance Variables**
- Defined inside a class’s methods (usually `__init__`) and belong to an instance of the class.
- **Example**:
  ```python
  class Dog:
      def __init__(self, name):
          self.name = name  # Instance variable

  dog = Dog("Buddy")
  print(dog.name)  # Output: Buddy

  ```

#### **b. Class Variables**
- Shared across all instances of a class, defined directly in the class body 
- **Example**:
  ```python
  class Dog:
      species = "Canis lupus"  # Class variable

      def __init__(self, name):
          self.name = name

  dog1 = Dog("Buddy")
  dog2 = Dog("Max")
  print(dog1.species, dog2.species)  # Output: Canis lupus Canis lupus
  ```

#### **c. Private Variables**
- Python uses naming conventions to indicate private variables (not strictly enforced)
- Single underscore (`_var`): Convention for "protected" variables (advisory, not enforced)
- Double underscore (`__var`): Name mangling to make variables harder to access outside the class(private)

- **Example**:
  ```python
  class Example:
      def __init__(self):
          self._protected_var = "Protected"
          self.__private_var = "Private"

      def get_private(self):
          return self.__private_var

  obj = Example()
  print(obj._protected_var)  # Output: Protected
  # print(obj.__private_var)  # Error: AttributeError
  print(obj.get_private())  # Output: Private
  print(obj._Example__private_var)  # Output: Private (name mangling)
  ```

#### **d. Special (Magic/Dunder) Variables**
- Variables with double underscores (e.g., `__name__`, `__doc__`) are used for special purposes in Python
- **Example**:
  ```python
  print(__name__)  # Output: __main__ (if run directly)
  ```

### **4. Variable Naming Conventions**
- **Rules**:
  - Must start with a letter (a-z, A-Z) or underscore (`_`).
  - Can contain letters, numbers (0-9), and underscores.
  - Case-sensitive (`var` and `Var` are different).
- **Conventions**:
  - Use `snake_case` for variable names (e.g., `my_variable`).
  - Use descriptive names (e.g., `user_age` instead of `x`).
  - Avoid reserved keywords (e.g., `class`, `for`, `if`).
  - Use uppercase for constants (e.g., `MAX_VALUE = 100`).

### **5. Variable Assignment and Mutability**
- **Assignment**: Variables are assigned using `=`. The type is inferred from the value.
  ```python
  x = 10  # x is an integer
  x = "Hello"  # x is now a string
  ```
- **Mutability**:
  - **Mutable types** (e.g., lists, dictionaries, sets): Changes affect the same object.
    ```python
    my_list = [1, 2, 3]
    my_list.append(4)
    print(my_list)  # Output: [1, 2, 3, 4]
    ```
  - **Immutable types** (e.g., integers, strings, tuples): Changes create a new object.
    ```python
    my_string = "hello"
    my_string = my_string + " world"
    print(my_string)  # Output: hello world
    ```

### **6. Type Hints for Variables**
Python supports optional type hints to indicate the expected type of a variable, improving code readability and enabling static type checking (e.g., with tools like `mypy`).

- **Example**:
  ```python
  age: int = 25
  name: str = "Alice"
  scores: list[int] = [90, 85, 88]

  def greet(person: str) -> str:
      return f"Hello, {person}!"

  print(greet(name))  # Output: Hello, Alice!
  ```

---

### **7. Dynamic Typing and Type Checking**
- **Dynamic Typing**: Variables can change types during execution.
  ```python
  var = 42  # Integer
  var = "Hello"  # Now a string
  ```
- **Type Checking**: Use `type()` or `isinstance()` to check a variable’s type.
  ```python
  var = 42
  print(type(var))  # Output: <class 'int'>
  print(isinstance(var, int))  # Output: True
  ```
