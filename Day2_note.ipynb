### 3. **Variable naming conventions, Constants and Literals**

#### **Variables** - A variable is a name used to store data in memory.        
             int a = 10;
             
             
- **Rules for creation of identifiers and variables:**

1. Must start with a letter (a-z, A-Z) or an underscore (_)
2. Can contain letters, digits (0-9), and underscores # my_var, _myVar, myVar1
3. Cannot contain spaces # my var
4. Cannot start with a digit # 2car
5. Cannot be a Python keyword # e.g., def, class, if, else, etc.
6. Cannot contain special characters (e.g., @, #, $, %, etc.)
7. Case-sensitive (e.g., myVar and myvar are different)

  **Constants:** - # Constants are variables that are not meant to change throughout the program. 
  - In Python, constants are typically defined using uppercase letters with underscores separating words.

  
          import constant
          print(constant.PI)  # Example of using a constant #3.14
          print(constant.gravity)  # Example of using a constant #9.8


  **Literals:** - Literals are fixed values that are represented directly in the code.
- They can be of various types, such as integers, floats, strings, and booleans.
- Integer literals are whole numbers, float literals are decimal numbers, string literals are sequences of characters
- and boolean literals represent truth values (True or False).

**Types of Literrals**
1. Numeric literals - Integers (e.g., 10, -5), Floats (e.g., 3.14, -0.001)
2. String literals - Single quotes ('Hello'), Double quotes ("World"), Triple quotes ('''Hello World''')
3. Boolean literals - True, False
4. None literal - Represents the absence of a value or a null value
5. Collection literals - Lists ([1, 2, 3]), Tuples ((1, 2, 3)), Sets ({1, 2, 3}), Dictionaries ({'key': 'value'})
6. Special literals - Ellipsis (…), NotImplemented
7. Bytes literals - b'Hello', bytearray(b'World')
8. Complex literals - Complex numbers (e.g., 3 + 4j)
9. Formatted string literals - f"Hello {name}", f"{value:.2f}"

### **6. Operators****

Python supports several types of operators for different operations.

**6.1 Arithmetic Operators**

These operators perform mathematical operations:

- `+` : Addition
- `-` : Subtraction
- `*` : Multiplication
- `/` : Division (Quotient with decimal values)
- `//` : Floor Division (integer result of division)
- `%` : Modulus (remainder)
- `**` : Exponentiation (power)


````python
x = 10

y = 5

print(x/y)#float
print(x//y) #int
print(x**y) 
print(x%y)
````

````python
x = 22

y = 20

x != y
````

**6.2 Comparison Operators**

These operators compare two values and return a Boolean (`True` or `False`):

- `==` : Equal
- `!=` : Not equal
- `>` : Greater than
- `<` : Less than
- `>=` : Greater than or equal to
- `<=` : Less than or equal to

````python
d=100
f=150

if d < f:
          print("d is greater than f")
````

````python
city = 'delhi'
population = 1.0

if population == 1.0:
    print("Population is 1 million")
if city == 'delhi' and population >= 2.0:
    print("City is Delhi and population is greater than 1.5 million")

# city == 'delhi' and population == 1.2
# city == 'delhi' or population == 1.1
````

````python
is_raining = False

if not is_raining:
    print("Dont go out its raining now")
elif is_raining or city == 'delhi':
    print("Take an umbrella if you're in delhi")
else:
    print("Its not raining")
````

**6.3 Logical Operators**

These operators combine multiple conditions:

- `and`: Returns `True` if both statements are true.
- `or`: Returns `True` if at least one statement is true.
- `not`: Reverses the result (True becomes False, False becomes True).
