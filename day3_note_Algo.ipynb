#### 3. Slicing Strings to Extract Substrings
##### Syntax: string[start:stop] → Extracts from index start up to (but not including) stop

```python
text = "Data Science"

# variable[1:6] - for indexing 

# print(text[0:n-1])
print(text[0:4])   # Output: Data
print(text[4:10])  # Output:Science
print(text[:4])    # Output: Data - it will print upto index 3
print(text[4:])    # Output: Science - it will print from index 4 to end
print(text[-7:-1]) # Output: Scienc
print(text[-10:])  # Output: DataScience
print(text[::2])   # Output: DtSine     
print(text[3:-1])
print(text[-1::-1]) # Reverse the string
print(text[::-1])  # Reverse the string
print(text[-1:-(len(text)+1):-1]) # Reverse the string
print(text[::3])   # Output: Daeie
print(text[1::2])  # Output: even index characters
print(text[0::2])  # Output: odd index characters

""" Note:

- Slicing does not include the end index
- Negative indices count from the end of the string
- The default start index is 0
- The default end index is the length of the string
- The default step is 1"""
```

```python
# Mini-project: sentence processing
from collections import Counter

sentence = input("Enter a sentence: ").strip()

# Split into words (splits on any whitespace; ignores extra spaces)
words = [w for w in sentence.split() if w]
word_count = len(words)

# Make each word start with a capital letter (preserves other letters)
title_words = [w[0].upper() + w[1:] if len(w) > 0 else w for w in words]
title_sentence = " ".join(title_words)

# Convert to list and back to string (vice versa)
as_list = words[:]                     # string -> list of words
as_string = " ".join(as_list)         # list -> single string (words joined by single space)

# Extra: most frequent word (if needed)
most_common = Counter([w.lower() for w in words]).most_common(1)
most_common_word = most_common[0][0] if most_common else None

print("Original sentence:", sentence)
print("Word count:", word_count)
print("Title-cased sentence:", title_sentence)
print("Words as list:", as_list)
print("List -> string:", as_string)
print("Most frequent word:", most_common_word)
```

```python
import mysql.connector
from collections import Counter
import csv

# MySQL connection setup (update credentials)
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="your_password",
    database="your_database"
)
cursor = conn.cursor()

# Ensure table exists
cursor.execute("""
CREATE TABLE IF NOT EXISTS sentence_info (
    id INT AUTO_INCREMENT PRIMARY KEY,
    sentence TEXT,
    word_count INT,
    title_case TEXT,
    word_list TEXT,
    unique_word_count INT,
    most_common_word TEXT
)
""")

# 1. Input sentence
sentence = input("Enter a sentence: ").strip()
words = sentence.split()
word_count = len(words)
title_sentence = sentence.title()
word_list = str(words)
unique_word_count = len(set(words))
most_common_word = Counter(words).most_common(1)[0][0] if words else None

# 2. Save to MySQL
cursor.execute("""
    INSERT INTO sentence_info (sentence, word_count, title_case, word_list, unique_word_count, most_common_word)
    VALUES (%s, %s, %s, %s, %s, %s)
""", (sentence, word_count, title_sentence, word_list, unique_word_count, most_common_word))
conn.commit()

print("Number of words:", word_count)
print("Title Case Sentence:", title_sentence)
print("List of words:", words)
print("Unique words:", unique_word_count)
print("Most frequent word:", most_common_word)

# 3. Retrieve all previous entries
print("\n--- Previous Entries ---")
cursor.execute("SELECT sentence, word_count, title_case, word_list, unique_word_count, most_common_word FROM sentence_info")
for row in cursor.fetchall():
    print(row)

# 4. Search sentences by keyword
keyword = input("\nEnter keyword to search: ").strip()
cursor.execute("SELECT sentence FROM sentence_info WHERE sentence LIKE %s", (f"%{keyword}%",))
results = cursor.fetchall()
print(f"Sentences containing '{keyword}':")
for r in results:
    print(r[0])

# 5. Export all data to CSV
with open("sentence_data.csv", "w", newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(["Sentence", "Word Count", "Title Case", "Word List", "Unique Word Count", "Most Common Word"])
    cursor.execute("SELECT sentence, word_count, title_case, word_list, unique_word_count, most_common_word FROM sentence_info")
    for row in cursor.fetchall():
        writer.writerow(row)
print("\nData exported to sentence_data.csv")

cursor.close()
conn.close() 
```
