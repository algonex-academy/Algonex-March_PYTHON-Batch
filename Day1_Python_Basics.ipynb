1. **Flow Charts**
- A flowchart is a diagram that visually shows the steps of a process or algorithm.

It uses shapes like:

Oval → Start/End

Rectangle → Process/Step

Diamond → Decision (Yes/No)

Parallelogram → Input/Output

```python
# Real Example: Checking if a number is even or odd

Flowchart steps:
Start
Input number
Check: number % 2 == 0
If Yes → print “Even”
If No → print “Odd”
End

**Flowcharts make it easy to understand the logic before writing code**
```

2. Algorithms

Real Example: Making Tea
Algorithm:

2. Add tea powder

4. Add sugar
5. Filter tea
6. Serve

This is how computers work — following exact steps

5.**Time & Space Complexity**

**These tell us how efficient an algorithm is**

A. **Time Complexity**
**It tells how much time an algorithm takes as input size increases**

Common time complexities:

| Complexity     | Meaning        | Example                 |
| -------------- | -------------- | ----------------------- |
| **O(1)**       | Constant time  | Accessing array element |
| **O(n)**       | Linear time    | Loop running n times    |
| **O(n²)**      | Quadratic time | Nested loops            |
| **O(log n)**   | Logarithmic    | Binary search           |
| **O(n log n)** | Fast sorting   | Merge sort, Quick sort  |

Real Example
- If you search your name in a list of 5 students, it’s fast
- If the list grows to 5,000 students, it takes longer

**Time complexity helps measure this difference**

```python
lsit = [1,2,3,4,5]
print(lsit) # 4ms
```

```python
lmt = [23,435,543,4,54,4544,656,456,4564,642,3,434,54,23,4543]
print(lmt) # 7ms
```

B. **Space Complexity**
- It tells how much memory/storage/space an algorithm needs

| Complexity   | Meaning                       |
| ------------ | ----------------------------- |
| **O(1)**     | Uses constant memory          |
| **O(n)**     | Memory grows with input size  |
| **O(log n)** | Very little extra memory used |

Real Example

- Storing 10 names needs small memory
- Storing 10,000 names needs much more

**Space complexity measures this requirement**

| Concept              | Meaning                           |

| **Pseudocode**       | Rough, English-like algorithm     |
| **Code**             | Actual programming implementation |
| **Time Complexity**  | How fast an algorithm runs        |
| **Space Complexity** | How much memory it needs          |

- Identifiers must start with a letter (a-z, A-Z) or an underscore (_), followed by letters, digits (0-9), or underscores.

     1. variable identifiers - (name = "Rahul")
     2. function identifiers - (def my_function():)

3. **Variable naming conventions, Constants and Literals**
             
             int a = 10
             
- Rules for creation of identifiers:
1. Must start with a letter (a-z, A-Z) or an underscore (_)
2. Can contain letters, digits (0-9), and underscores # my_var, _myVar, myVar1
3. Cannot contain spaces # my var
4. Cannot start with a digit # 2car
5. Cannot be a Python keyword # e.g., def, class, if, else, etc.
6. Cannot contain special characters (e.g., @, #, $, %, etc.)
7. Case-sensitive (e.g., myVar and myvar are different)
