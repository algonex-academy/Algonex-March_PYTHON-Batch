## 4. Nested Conditionals
#### You can put an if statement inside another if or else.


if condition1:

          if condition2:

                    # nested block

          else:

                    # nested else


```python
age = int(input("Enter your age: "))
citizen = None                  # bool(input("Are you a citizen? (only for True): "))

if age >= 18:
    if citizen: # Since citizen is None, it will be treated as False
        print("You can vote")
    else:
        print("Only citizens can vote")
else:
    print("You are underage")
```

### Nested if condition

```python
'''
/ - Division - Float Value
// - Floor Division - Integer Value
% - Modulus - Remainder Value
'''
number = int(input("Enter a number: "))  # int and flaot values we can enter

if number % 2 == 0:
    print("Number is even")
    if number > 0:
        print(f"{number} is positive even")
        if number > 10000:
            print("It's a big positive even number")
        
        else:
            print(f"{number} is small positive even number")
    else:
        print(f"{number} is negative even number")
    
else:
    print("Number is odd")
```

## **Nested if-else**
### Let me first show you how to use these four conditions to make a decision using nested if-else statement.

```python
marks_obt = float(input('Enter Total Marks You Have Secured: '))
if marks_obt >=50 and marks_obt <60:
    print('You Have Got SECOND division')
else:
    if marks_obt >=60 and marks_obt <75:
        print('You Have Passed With FIRST division', end='\n')
    else:
        if marks_obt >=75 and marks_obt <90:
            print('You Have Got FIRST division with DISTINCTION', end='\n\n')
        else:
            if marks_obt >= 90:
                print('Congratulation! OUTSTANDING PERFORMANCE', end='\n\n')
            else:
                print("Sorry! You Have FAILED!", end='\n\n')
print('Good Luck For Your Future')
```

## 5. Ternary Operator (Short if-else)
#### => Used for short, one-line decisions.

### Syntax:

#### value_if_true if condition else value_if_false

```python
"Kiran is sitting" if name == "Kiran" else "Kiran is not sitting"
```

```python
# Example:
# Ternary Operator Example
age = int(input("Enter your age: "))
result = "Adult" if age >= 18 else "Minor"
print(result)
```

```python
if age >= 18:
    print("You are an adult") 
else:
    print("You are a minor")
```

```python

| Statement Type       | Purpose                       | Syntax Example                    |
| -------------------- | ----------------------------- | --------------------------------- |
| `if`                 | Check one condition           | `if a > b:`                       |
| `elif`               | Check additional conditions   | `elif a == b:`                    |
| `else`               | Run when no condition is True | `else:`                           |
| **Nested if**        | `if` inside another `if`      | `if a > 0: if a < 10:`            |
| **Ternary Operator** | One-line `if-else`            | `"Even" if n % 2 == 0 else "Odd"` |
```
