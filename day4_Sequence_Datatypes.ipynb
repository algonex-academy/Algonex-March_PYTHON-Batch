#### **Sets** {} -- Curly Brackets  
A set is an unordered collection of unique items. Sets do not allow duplicate elements.
- we can store diff data types  
- it automatically eleminate the duplcates and output values will sort in ascending order  by default

```python
my_set1 = {1, 2,23,2 ,6,34,3432,11,234,4}
my_set2 = {4, 15, 6}
my_set2.pop() # removes and returns an arbitrary(first) element
# my_set1.find(23) # AttributeError: 'set' object has no attribute 'find'
my_set1.add(999) 
my_set1.remove(23) # removes specified element
my_set2.insert(2,60) # AttributeError
# my_set1.count(234) # AttributeError: 'set' object has no attribute 'count'
my_set2.append(9023) # AttributeError
my_set1.index(234) # no attribute 'index' in set
my_set1
```

- Intersection (&): Returns common elements.  
- Difference (-): Returns elements in one set but not the other.
- Symmetric Difference (^): Returns elements in either set but not both.

```python

"""
**Project 1: Grocery List Manager**

Objective:
Create a simple grocery list manager where users can add items, view the list, and remove duplicates (using set functionality).

Steps:
Ask the user to input grocery items (comma-separated).  
Store these items in a list.  
Convert the list to a set to remove duplicates.  
Display the updated grocery list.  
"""

# asking for items

intems = input("Please enter items name separating by commas :")

# item convered into list
intems_list = intems.strip().split(',') # ['Banana', ' Beans', ' APPLE Ginger']

# print(intems_list)
# # type casting
intems_unique = set(intems_list)

item_unique_list = list(intems_unique)

# # Show unique items list
print(f"Thank you for entering items here is your unique list of items {item_unique_list}")

# # end of function
# # patoto, tamato, chilli, sauces, bread, milk, milk, milk, sauces, sauces
```

```python
""" Mini Project: Grocery List Manager

Objective:
Allow the user to:
Add items
Remove items
View list
Exit """ 

# Grocery List Manager
grocery_list = []

while True:
    print("\n1. Add Item\n2. Remove Item\n3. View List\n4. Exit")
    choice = input("Enter your choice: ")

    if choice == '1':
        item = input("Enter item to add: ")
        grocery_list.append(item)
        print(f"{item} added.")
    elif choice == '2':
        item = input("Enter item to remove: ")
        if item in grocery_list:
            grocery_list.remove(item)
            print(f"{item} removed.")
        else:
            print("Item not found.")
    elif choice == '3':
        print("Your Grocery List:", grocery_list)
    elif choice == '4':
        print("Exiting Grocery List Manager. Goodbye!")
        break
    else:
        print("Invalid choice.")
```
