**Dictionary in python**  
A dictionary in Python is an unordered, mutable, and indexed collection of key-value pairs.

### A word (the key) maps to a meaning (the value).
 Each key is unique, and it maps to a corresponding value. 
  
Key Characteristics:  
- Unordered
- Mutable   (Changeable) # You can add, remove, or update items in a dictionary after it is created.
- Key must be unique
- Values can be any type

```python
student = {
    "name": "Sara",
    "age": 22,
    "course": "B.Tech",
    "year": [2023,2024,25,2026],
    "subjects": ["Maths", "Physics", "Chemistry"],
    ["Location1","location2"]: ["Hyderabad", "Bangalore"]
}

""" In Python, a dictionary is a collection of key-value pairs.
Each key is unique and is used to access its corresponding value.
For example, in the dictionary above:

"name" is the key → "Sara" is the value.
"age" is the key → 22 is the value.""" 

type(student)
student["name"] = "Pravs" #modifying existing key
student["age"] = 23 #Indexing existing key
student["course"] = "M.Tech"
print(student)
```

#### Common Operations with Dictionary

| Operation       | Code Example           | Description                   |
| --------------- | ---------------------- | ----------------------------- |
| Create          | `d = {"a": 1, "b": 2}` | Create dictionary             |
| Access value    | `d["a"]`, `d.get(key)` | Get value of key "a"          |
| Update value    | `d["a"] = 10`          | Change value of key "a"       |
| Add key-value   | `d["c"] = 3`           | Add new entry                 |
| Delete key      | `del d["b"]`           | Remove key "b"                |
| Keys only       | `d.keys()`             | Get list of all keys          |
| Values only     | `d.values()`           | Get list of all values        |
| All items       | `d.items()`            | Get key-value pairs as tuples |
| Check existence | `"a" in d`             | Returns `True` if key exists  |

| **Method**       | **Description**                                                                                     |
|------------------|-----------------------------------------------------------------------------------------------------|
| `dict.get(key)`  | Returns the value for a given key; if the key does not exist, returns a default value (e.g., `None`).|
| `dict.keys()`    | Returns a view of all keys in the dictionary.                                                        |
| `dict.values()`  | Returns a view of all values in the dictionary.                                                      |
| `dict.items()`   | Returns a view of all key-value pairs in the dictionary as tuples.                                   |
| `dict.update()`  | Updates the dictionary with elements from another dictionary or iterable of key-value pairs.         |
| `dict.clear()`   | Removes all elements from the dictionary.                                                           |
| `dict.pop(key)`  | Removes the key-value pair and returns the value; raises a `KeyError` if the key doesn't exist.      |

```python
# Student Report Card Management System
# Using Dictionaries Only

students = {}   # Main dictionary to store all student data

def add_student():
    roll = input("Enter Roll Number: ")
    name = input("Enter Student Name: ")

    print("\nEnter Marks:")
    m1 = int(input("Math: "))
    m2 = int(input("Science: "))
    m3 = int(input("English: "))

    total = m1 + m2 + m3
    percentage = total / 3

    students[roll] = {
        "name": name,
        "marks": {"Math": m1, "Science": m2, "English": m3},
        "total": total,
        "percentage": percentage
    }

    print("\nStudent added successfully!\n")


def search_student():
    roll = input("Enter Roll Number to Search: ")

    if roll in students:
        s = students[roll]
        print("\n--- Student Found ---")
        print("Name:", s["name"])
        print("Marks:", s["marks"])
        print("Total:", s["total"])
        print("Percentage:", round(s["percentage"], 2))
        print("---------------------\n")
    else:
        print("\nStudent Not Found!\n")


def display_all():
    if not students:
        print("\nNo records available.\n")
        return

    print("\n------ ALL STUDENT REPORT CARDS ------")
    for roll, data in students.items():
        print(f"\nRoll No: {roll}")
        print(f"Name: {data['name']}")
        print("Marks:", data["marks"])
        print("Total:", data["total"])
        print("Percentage:", round(data["percentage"], 2))
    print("--------------------------------------\n")


def main_menu():
    while True:
        print("===== Student Report Card System =====")
        print("1. Add Student")
        print("2. Search Student")
        print("3. Display All Students")
        print("4. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_student()
        elif choice == "2":
            search_student()
        elif choice == "3":
            display_all()
        elif choice == "4":
            print("Exiting...")
            break
        else:
            print("Invalid choice. Try Again!\n")


# Run the Project
main_menu()
```
